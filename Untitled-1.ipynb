```python
while True:
    try:
        a=int(input("enter an num:"))
        b=int(input("enter an num:"))
        sym=input("enter an operator (+,-,*,/):")
        if sym=='+':
            print(a+b)
        elif sym=='-':
            print(a-b)
        elif sym=='*':
            print(a*b)
        elif sym=='/':
            print(a/b)
        else:
            raise Exception("invalid operator")
        break
    except Exception as e:
        print("error occurs due to :",e)
    else:
        print("no error occurs")
    finally:
        print("end of program")
```
